1. What is a token in text processing/NLP?

A token is a broken-down version of a word or phrase that is given a numerical value, which is then processed by a large language model to provide contextual meaning.  

2. Why is embedding necessary for LLMs like BERT, GPTs? 

The token, which is processed by the large language model, is placed inside of a vector embedding that provides a “score” or mathematical representation of the contextual meaning of the token. This is how LLMs are able to know the difference between the meanings of words. 

3. Code Below

```python
from transformers import AutoTokenizer, AutoModel
import torch
import torch.nn.functional as F

#Using Apple Computer
device = "mps"

#Load model and tokenizer for BERT
tokenizer = AutoTokenizer.from_pretrained("google-bert/bert-base-uncased")
model = AutoModel.from_pretrained("google-bert/bert-base-uncased").to(device)
```

```python
par_1 = """Masked language modeling (MLM): taking a sentence, the model randomly masks 15% of the words in the input then run the entire masked sentence 
            through the model and has to predict the masked words. This is different from traditional recurrent neural networks (RNNs) that usually see 
            the words one after the other, or from autoregressive models like GPT which internally masks the future tokens. 
            It allows the model to learn a bidirectional representation of the sentence."""

par_2 = """Next sentence prediction (NSP): the models concatenates two masked sentences as inputs during pretraining. 
            Sometimes they correspond to sentences that were next to each other in the original text, sometimes not. 
            The model then has to predict if the two sentences were following each other or not."""
```

```python
inputs_1 = tokenizer(par_1, return_tensors="pt", padding=True, truncation=True).to(device)
inputs_2 = tokenizer(par_2, return_tensors="pt", padding=True, truncation=True).to(device)

# Print tokens for both paragraphs
print("Tokens P1:", tokenizer.convert_ids_to_tokens(inputs_1['input_ids'][0]))
print("\nTokens P2:", tokenizer.convert_ids_to_tokens(inputs_2['input_ids'][0]))
```

```python

#Get embeddings
with torch.no_grad():
    outputs_1 = model(**inputs_1)
    outputs_2 = model(**inputs_2)

emb_1 = outputs_1.last_hidden_state[:, 0, :]
print("\nEmbedding Vector P1:", emb_1)
```

```python
emb_2 = outputs_2.last_hidden_state[:, 0, :]
print("\nEmbedding Vector P2:", emb_2)
```

```python
#Cosine similarity
similarity = F.cosine_similarity(emb_1, emb_2)
print("\nCosine Similarity:", similarity.item())
```
